## 3. Initialisation du TTS

Charge les modeles et extrait le speaker embedding depuis `audio.wav`.

```python
from tts import TTS

tts = TTS(audio_path="audio.wav")
```

## 4. Generation

Parametres pour reduire le grincement :
- `temperature=0.5` : plus stable, moins d'artefacts
- `num_beams=16` : recherche plus large
- `repetition_penalty=1.5` : evite les motifs repetitifs

```python
tts.play("aksel waaaaaay", temperature=0.5, num_beams=16, repetition_penalty=1.5)
```

```python
tts.play("Bonjour, bienvenue dans le modele de synthese vocale.", temperature=0.5, num_beams=16, repetition_penalty=1.5)
```

```python
tts.play("damaa beuga dem sama ker wayé yoon bi dafa meti", temperature=0.8, num_beams=16, repetition_penalty=1.5)
```

## 5. Sauvegarder un audio

```python
tts.generate("Bopou louminaa dafa reuy", output_path="output.wav", temperature=0.8, num_beams=16, repetition_penalty=1.5)
```

```python
tts.generate("dama beuga dem sama ker", output_path="output.wav", temperature=0.8, num_beams=16, repetition_penalty=1.5)
```
