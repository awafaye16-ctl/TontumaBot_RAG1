### 1.1 Configuration du HF_TOKEN

Pour définir votre `HF_TOKEN` et éviter les avertissements de Hugging Face :

1.  **Obtenez un jeton (token) Hugging Face :**
    *   Rendez-vous sur [Hugging Face Settings/Tokens](https://huggingface.co/settings/tokens).
    *   Créez un nouveau jeton d'accès avec le rôle `Read` (lecture).

2.  **Ajoutez le jeton aux secrets de Colab :**
    *   Dans Colab, cliquez sur l'icône "🔑" (Secrets) dans le panneau de gauche.
    *   Cliquez sur `+ New secret`.
    *   Pour le champ `Name`, entrez `HF_TOKEN` (cela doit correspondre au nom que le code cherchera).
    *   Pour le champ `Value`, collez le jeton que vous avez copié depuis Hugging Face.
    *   Assurez-vous que l'option `Notebook access` est cochée pour ce secret.

3.  **Chargez le jeton dans votre code :**
    *   La cellule suivante montre comment charger le `HF_TOKEN` depuis les secrets de Colab. Une fois chargé, la bibliothèque `transformers` l'utilisera automatiquement pour vos requêtes.

```python
from google.colab import userdata

# Charge le HF_TOKEN depuis les secrets de Colab
HF_TOKEN = userdata.get('HF_TOKEN')

# Vous pouvez maintenant utiliser HF_TOKEN directement ou la bibliothèque
# transformers le détectera automatiquement une fois qu'il sera chargé dans l'environnement.
# Si vous voulez le définir explicitement pour la session courante:
import os
os.environ['HF_TOKEN'] = HF_TOKEN

print("HF_TOKEN chargé avec succès depuis les secrets de Colab.")
```
